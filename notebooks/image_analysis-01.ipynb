```python
from PIL import Image

dimension_errors = []

for _, row in df.iterrows():
    image_path = os.path.join(DATASET_DIR, row["image"])
    black_path = os.path.join(DATASET_DIR, row["black"])
    label_path = os.path.join(DATASET_DIR, row["label"])

    with Image.open(image_path) as img:
        image_size = img.size

    with Image.open(black_path) as img:
        black_size = img.size

    with Image.open(label_path) as img:
        label_size = img.size

    if image_size != black_size or image_size != label_size:
        dimension_errors.append({
            "name": row["name"],
            "image": image_size,
            "black": black_size,
            "label": label_size
        })

print("Total checked:", len(df))
print("Dimension mismatches:", len(dimension_errors))

if dimension_errors:
    for error in dimension_errors[:10]:
        print(error)
else:
    print("All image, black, and label dimensions match!")
```

```python
import cv2
import numpy as np
from PIL import Image

row = df.iloc[0]

label_path = os.path.join(DATASET_DIR, row["label"])
mask = np.array(Image.open(label_path))

# Create a binary mask for pixels with value 255
binary_mask = (mask == 255).astype(np.uint8)

# Detect connected components
num_labels, components, stats, centroids = cv2.connectedComponentsWithStats(
    binary_mask,
    connectivity=8
)

# Exclude background (component 0)
component_areas = stats[1:, cv2.CC_STAT_AREA]

print("Image:", row["name"])
print("Number of connected regions:", num_labels - 1)
print("Largest region area:", component_areas.max())
print("Smallest region area:", component_areas.min())
print("Average region area:", component_areas.mean())
```

```python
results = []

for _, row in df.iterrows():
    label_path = os.path.join(DATASET_DIR, row["label"])
    mask = np.array(Image.open(label_path))

    binary_mask = (mask == 255).astype(np.uint8)

    num_labels, components, stats, centroids = (
        cv2.connectedComponentsWithStats(
            binary_mask,
            connectivity=8
        )
    )

    areas = stats[1:, cv2.CC_STAT_AREA]

    results.append({
        "name": row["name"],
        "class": row["class"],
        "split": row["split"],
        "num_components": len(areas),
        "largest_area": areas.max() if len(areas) else 0,
        "smallest_area": areas.min() if len(areas) else 0,
        "average_area": areas.mean() if len(areas) else 0
    })

morphology_df = pd.DataFrame(results)

print(morphology_df.head())
print("\nTotal masks analyzed:", len(morphology_df))
```

```python
row = df.iloc[0]

label_path = os.path.join(DATASET_DIR, row["label"])
mask = np.array(Image.open(label_path))

binary_mask = (mask == 255).astype(np.uint8)

kernel = np.ones((3, 3), np.uint8)

opening = cv2.morphologyEx(
    binary_mask,
    cv2.MORPH_OPEN,
    kernel
)

closing = cv2.morphologyEx(
    binary_mask,
    cv2.MORPH_CLOSE,
    kernel
)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(binary_mask, cmap="gray")
axes[0].set_title("Original Binary Mask")

axes[1].imshow(opening, cmap="gray")
axes[1].set_title("After Opening")

axes[2].imshow(closing, cmap="gray")
axes[2].set_title("After Closing")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
def analyze_mask(binary_mask):
    num_labels, labels, stats, centroids = (
        cv2.connectedComponentsWithStats(
            binary_mask,
            connectivity=8
        )
    )

    areas = stats[1:, cv2.CC_STAT_AREA]

    return {
        "Components": len(areas),
        "Total foreground pixels": int(binary_mask.sum()),
        "Largest area": int(areas.max()) if len(areas) else 0,
        "Smallest area": int(areas.min()) if len(areas) else 0
    }


print("Original:", analyze_mask(binary_mask))
print("Opening:", analyze_mask(opening))
print("Closing:", analyze_mask(closing))
```

```python
def calculate_iou(mask1, mask2):
    intersection = np.logical_and(mask1, mask2).sum()
    union = np.logical_or(mask1, mask2).sum()

    return intersection / union if union > 0 else 1.0


print("Opening IoU:", calculate_iou(binary_mask, opening))
print("Closing IoU:", calculate_iou(binary_mask, closing))
```

```python
kernel_sizes = [3, 5, 7]

for size in kernel_sizes:
    kernel = np.ones((size, size), np.uint8)

    opened = cv2.morphologyEx(
        binary_mask,
        cv2.MORPH_OPEN,
        kernel
    )

    closed = cv2.morphologyEx(
        binary_mask,
        cv2.MORPH_CLOSE,
        kernel
    )

    print(f"\nKernel Size: {size}x{size}")

    print("Opening IoU:",
          calculate_iou(binary_mask, opened))

    print("Closing IoU:",
          calculate_iou(binary_mask, closed))
```
